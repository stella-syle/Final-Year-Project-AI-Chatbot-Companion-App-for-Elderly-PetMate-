# 1. Import data

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

#read csv
testBefore = pd.read_excel('testBefore.xlsx', sheet_name='Sheet1')

#change format
testBefore['response_sec'] = pd.to_timedelta(testBefore['response sec(around)'].astype(str)).dt.total_seconds()
testBefore
```

```python
testBefore = testBefore.drop(columns=['response sec(around)'])
testBefore
```

```python
testAfter = pd.read_excel('testAfter.xlsx', sheet_name='Sheet1')
testAfter['response_sec'] = pd.to_timedelta(testAfter['response sec(around)'].astype(str)).dt.total_seconds()
testAfter = testAfter.drop(columns=['response sec(around)'])

testAfter
```

# 2. Analysis for response_sec

```python
testBefore['response_sec'].describe()
```

```python
testAfter['response_sec'].describe()
```

```python
before_mean = testBefore['response_sec'].mean()
before_max = testBefore['response_sec'].max()

after_mean = testAfter['response_sec'].mean()
after_max = testAfter['response_sec'].max()

compare = pd.DataFrame({
    'Before': [before_mean, before_max],
    'After': [after_mean, after_max]
}, index=['Mean Response Time', 'Max Response Time'])

compare['Increase %'] = ((compare['After'] - compare['Before']) / compare['Before']) * 100

print(compare.round(2))
```

```python
compare[['Before','After']].plot(kind='bar')

plt.title('Before vs After Response Time Comparison')
plt.ylabel('Seconds')
plt.xticks(rotation=0)
plt.show()
```

- Mean Response Time increase 66%
- Max Response Time increase 51%
- The system now takes longer to generate outputs, especially for complex or lengthy inputs

# 3. Correlation among text length and response time

```python
testBefore['Length'] = testBefore['content'].str.len()
testAfter['Length'] = testAfter['content'].str.len()

# clacu correlation
before_corr = testBefore['Length'].corr(testBefore['response_sec'])
after_corr = testAfter['Length'].corr(testAfter['response_sec'])

print("Correlation coefficient (Before):", round(before_corr, 3))
print("Correlation coefficient (After):", round(after_corr, 3))

```

```python
import seaborn as sns

plt.figure()

testBefore['Length'] = testBefore['content'].str.len()
testAfter['Length'] = testAfter['content'].str.len()

sns.regplot(
    x = testBefore['Length'],
    y = testBefore['response_sec']
)

plt.title('Before: Length of Content vs Response time(sec)')
plt.xlabel('Content Length')
plt.ylabel('Response time(sec)')

plt.tight_layout()
plt.show()
```

```python
plt.figure()

sns.regplot(
    x = testAfter['Length'],
    y = testAfter['response_sec']
)

plt.title('After:  Length of Content vs Response time(sec)')
plt.xlabel('Content Length')
plt.ylabel('Response time(sec)')

plt.tight_layout()
plt.show()
```

Before Prompt (r = -0.082)

- Input length did not influence system speed.
- Latency appeared random, driven by internal generation factors (e.g., token sampling, temperature randomness).

After Prompt (r = 0.501)
- Moderate positive correlation
- Longer content > longer response times

After the adjusting, generate time start dominated by token processing.

# 4. How new prompt affect response time for same content 

```python
# merge same content
merged = pd.merge(
    testBefore[['content','response_sec']],
    testAfter[['content','response_sec']],
    on='content',
    suffixes=('_Before','_After'))

# Find difference
merged['Difference'] = merged['response_sec_After'] - merged['response_sec_Before']

merged
```

0 and 1 is correct, since testBefore has one 你好 with 25.32 sec and testAfter has 2 你好 with both 47.42 sec

```python
content_diff = merged.groupby('content')['Difference'].mean()
```

```python
import matplotlib.font_manager as fm

font_path = "/System/Library/Fonts/STHeiti Light.ttc"

font_prop = fm.FontProperties(fname=font_path)

plt.rcParams['font.family'] = font_prop.get_name()
plt.rcParams['axes.unicode_minus'] = False


merged['Increase_%'] = merged['Difference'] / merged['response_sec_Before'] *100

pct = merged.groupby('content')['Increase_%'].mean()

pct.plot(kind='bar')

plt.title('Latency Increase % (After vs Before)')
plt.ylabel('%')
plt.show()
```

Overall latency increased across all contents after the change

# 5. Latency Increase per Token 

```python
import statsmodels.api as sm

X_before = testBefore['Length']
y_before = testBefore['response_sec']

X_before = sm.add_constant(X_before)

model_before = sm.OLS(y_before, X_before).fit()

print("Beforethe adjustment:")
print(model_before.summary())
```

```python
X_after = testAfter['Length']
y_after = testAfter['response_sec']

X_after = sm.add_constant(X_after)

model_after = sm.OLS(y_after, X_after).fit()

print("After the adjustment:")
print(model_after.summary())
```

```python
before_beta = model_before.params['Length']
after_beta  = model_after.params['Length']

print("Latency increase per token (Before):",
      round(before_beta,3),"sec")

print("Latency increase per token (After):",
      round(after_beta,3),"sec")
```

After prompt tuning, response latency increased by approximately 1.82 seconds per input token, compared to 0.12 seconds per token before tuning, suggesting inference time has become significantly more input-length dependent.

# 6. Analysis for Abnormal response

```python
# count abnormal response
before_count = (testBefore['Abnormal response'] == 'Y').sum()
after_count = (testAfter['Abnormal response'] == 'Y').sum()

# decrease ratio
decrease_ratio = (before_count - after_count) / before_count * 100

abnormal_table = pd.DataFrame({
    'Before': [before_count, before_count / len(testBefore) * 100],
    'After': [after_count, after_count / len(testAfter) * 100],
    'Change': [before_count - after_count, decrease_ratio]
}, index=['Abnormal Count', 'Abnormal Rate (%)'])

print(abnormal_table.round(2))
```

```python
ax = abnormal_table.T.plot(kind='bar', figsize=(8,6))
plt.title('Abnormal Response Comparison')
plt.ylabel('Value')
plt.xticks(rotation=0)
plt.legend


plt.tight_layout()
plt.show()
```

- Abnormal Count decreased from 3 to 1, 66% decline.
- The intervention effectively reduced both the problem of abnormal responses.

The research reveals a considerable improvement in aberrant responses following the intervention. The aberrant count reduced from three to one, resulting in a two-case reduction. More crucially, the anomalous rate fell from 27.27% to 10.00%, representing a 66.67% fall. This suggests that the intervention was successful in reducing aberrant responses in both absolute numbers and relative proportions.
